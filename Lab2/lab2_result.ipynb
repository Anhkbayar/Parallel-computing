```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('results.csv')
print(df)
```

```python
workloads = sorted(df["Workload"].unique())

for w in workloads:
    subset = df[df["Workload"] == w]

    plt.figure()
    plt.plot(subset["Tasks"], subset["Serial"], marker='o', label="Serial")
    plt.plot(subset["Tasks"], subset["Parallel"], marker='o', label="Parallel Spawn")
    plt.plot(subset["Tasks"], subset["Spinning"], marker='o', label="Thread Pool Spinning")
    plt.plot(subset["Tasks"], subset["Sleeping"], marker='o', label="Thread Pool Sleeping")

    plt.xlabel("Number of Tasks")
    plt.ylabel("Execution Time (seconds)")
    plt.title(f"Task System Performance (Workload = {w})")
    plt.legend()
    plt.grid(True)

    plt.show()
```
