```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("output.csv")
df.head()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(df["position"], df["temperature"], marker='')

plt.xlabel("Position")
plt.ylabel("Temperature")
plt.title("Temperature Distribution")
plt.grid(True)

plt.show()
```
