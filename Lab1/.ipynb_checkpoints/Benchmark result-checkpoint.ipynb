```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('results.csv')
print(df)
# print(df.to_string())
```

```python
df_clean = df[df["run_i"] > 3]
print(df_clean)
# print(df_clean.to_string())
```

```python
summary = (df_clean.groupby(["workload", "impl", "threads"]).agg(
        mean_ms=("elapsed_ms", "mean"),
        p95_ms=("elapsed_ms", lambda x: x.quantile(0.95))
    )
    .reset_index()
)

# print(summary)
print(summary.to_string())
```

```python
thread_serial_sum = summary.loc[
    (summary["workload"] == "sum") &
    (summary["impl"] == "threads") &
    (summary["threads"] == 1),
    "mean_ms"
].iloc[0]
thread_serial_equation = summary.loc[
    (summary["workload"] == "equation") &
    (summary["impl"] == "threads") &
    (summary["threads"] == 1),
    "mean_ms"
].iloc[0]
omp_serial_sum = summary.loc[
    (summary["workload"] == "sum") &
    (summary["impl"] == "OMP") &
    (summary["threads"] == 1),
    "mean_ms"
].iloc[0]
omp_serial_equation = summary.loc[
    (summary["workload"] == "equation") &
    (summary["impl"] == "OMP") &
    (summary["threads"] == 1),
    "mean_ms"
].iloc[0]
print(thread_serial_sum, thread_serial_equation)
print(omp_serial_sum, omp_serial_equation)

baseline = {
    ("sum", "threads"): thread_serial_sum,
    ("equation", "threads"): thread_serial_equation,
    ("sum", "OMP"): omp_serial_sum,
    ("equation", "OMP"): omp_serial_equation,
}
summary["speedup"] = None

for (workload, impl), T1 in baseline.items():
    mask = (
        (summary["workload"] == workload) &
        (summary["impl"] == impl)
    )
    summary.loc[mask, "speedup"] = T1 / summary.loc[mask, "mean_ms"]

summary["efficiency"] = None
summary["efficiency"] = summary["speedup"] / summary["threads"]

print(summary)
```

```python
omp = summary[(summary["impl"] == "OMP") & (summary["workload"] == "sum")]
threads = summary[(summary["impl"] == "threads") & (summary["workload"] == "sum")]
serial = summary[(summary["impl"] == "serial") & (summary["workload"] == "sum")]

plt.figure()

plt.plot(omp["threads"], omp["mean_ms"], marker='o', label="OpenMP")
plt.plot(threads["threads"], threads["mean_ms"], marker='o', label="std::thread")
plt.plot(serial["threads"], serial["mean_ms"], marker='o', label="Serial")

plt.xlabel("Threads")
plt.ylabel("Elapsed time (ms)")
plt.title("Sum workload")
plt.legend()
plt.grid(True)

plt.show()

print(omp)
```

```python
fig,axs = plt.subplots(1,2, figsize=(10,4))

axs[0].plot(omp["threads"], omp["speedup"], marker='o', label="OpenMP")
axs[0].plot(threads["threads"], threads["speedup"], marker='o', label="std::thread")
axs[0].set_xlabel("threads")
axs[0].set_ylabel("speedup")
axs[0].set_title("Sum speedup")
axs[0].grid()
axs[0].legend()

axs[1].plot(omp["threads"], omp["efficiency"], marker='o', label="OpenMP")
axs[1].plot(threads["threads"], threads["efficiency"], marker='o', label="std::thread")
axs[1].set_title("Sum efficiency")
axs[1].set_xlabel("threads")
axs[1].set_ylabel("speedup")
axs[1].grid()
axs[1].legend()

plt.tight_layout()
```

```python
omp = summary[(summary["impl"] == "OMP") & (summary["workload"] == "equation")]
threads_impl = summary[(summary["impl"] == "threads") & (summary["workload"] == "equation")]
serial = summary[(summary["impl"] == "serial") & (summary["workload"] == "equation")]

plt.figure()

plt.plot(omp["threads"], omp["mean_ms"], marker='o', label="OpenMP")
plt.plot(threads_impl["threads"], threads_impl["mean_ms"], marker='o', label="std::thread")
plt.plot(serial["threads"], serial["mean_ms"], marker='o', label="Serial")

plt.xlabel("Threads")
plt.ylabel("Elapsed time (ms)")
plt.title("Equation workload")
plt.legend()
plt.grid(True)

plt.show()
```
